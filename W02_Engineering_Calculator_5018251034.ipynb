```python
# Task 1: Speed Unit Conversion
speed_kmh = float(input("Enter speed [km/h]: "))
speed_ms = speed_kmh / 3.6
print(f"Speed = {speed_ms:.2f} m/s")
```

```python
# Task 2: Power Unit Conversion
power_kw = float(input("Enter Power [kW]: "))
power_w = 1000 * power_kw
print(f"Power = {power_w:.2f} W")
```

```python
# Task 3: Electrical Power and Energy Calculator
# Test 1
voltage = float(input("Enter voltage [V]: "))
current = float(input("Enter current [A]: "))
operating_time = float(input("Enter operating time [hours]: "))
power = voltage * current
energy = power * operating_time
print("===== ELECTRICAL CALCULATION =====")
print(f"Voltage = {voltage:.2f} V")
print(f"Current = {current:.2f} A")
print(f"Power = {power:.2f} W")
print(f"Operating Time = {operating_time:.2f} hour")
print(f"Energy = {energy:.2f} Wh")
```

```python
# Test 2
v2 = float(input("Enter voltage [V]: "))
i2 = float(input("Enter current [A]: "))
t2 = float(input("Enter operating time [hours]: "))
p2 = v2 * i2
e2 = p2 * t2

print("===== ELECTRICAL CALCULATION =====")
print(f"Voltage = {v2:.2f} V")
print(f"Current = {i2:.2f} A")
print(f"Power = {p2:.2f} W")
print(f"Operating Time = {t2:.2f} hour")
print(f"Energy = {e2:.2f} Wh")
```

```python
# Task 4: Comparison and Logical Operators
test_cases = [
    {"test": "Test 1", "voltage": 12.0, "current": 5.0, "time": 2.0},
    {"test": "Test 2", "voltage": 24.0, "current": 8.5, "time": 3.0},
    {"test": "Test 3", "voltage": 48.0, "current": 10.0, "time": 1.5},
]
power_limit = 200
voltage_limit = 24

for data in test_cases:
    voltage = data["voltage"]
    current = data["current"]
    time_hours = data["time"]

    power = voltage * current
    energy = power * time_hours
    is_high_power = power > power_limit
    is_high_voltage = voltage > voltage_limit
    print(f"===== ELECTRICAL CALCULATION ({data['test']}) =====")
    print(f"Voltage = {voltage:.2f} V")
    print(f"Current = {current:.2f} A")
    print(f"Power = {power:.2f} W")
    print(f"Operating Time = {time_hours:.2f} hour")
    print(f"Energy = {energy:.2f} Wh")
    print(f"High power : {is_high_power}")
    print(f"High voltage : {is_high_voltage}")
    print(f"Both conditions : {is_high_power and is_high_voltage}")
    print(f"Any condition : {is_high_power or is_high_voltage}\n")
```
